```python
personas = {
    "nombre" : ["Juan","Jaime","Raul"],
    "apellido" : ["Fernández","Fernández","Martínez"],
    "email" : ["JR@gmail.com","JF@gmail.com","RM@gmail.com"]
}
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(personas)
```

```python
df
```

```python
filtro = (df["nombre"] == "Jaime") & (df['apellido'] == 'Fernández')
```

```python
df.loc[filtro,'email'] #Nos permite seleccionar filas y columnas en una única operación
```

```python
filtro_2 = (df["nombre"] == "Jaime") | (df["nombre"] == "Juan")
```

```python
df.loc[filtro_2]
```

```python
df.loc[~filtro_2] #Buscamos lo contrario a nuestro filtro
```
